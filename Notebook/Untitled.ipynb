```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_csv("../cleaned_data/IoT_cleaned.csv")
```

```python
df.head()
```

```python
print(df.shape)
```

```python
df.info()
```

```python
df.describe().T
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="Congestion_Level"
)

plt.title("Congestion Level Distribution")
plt.xlabel("Congestion Level")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Vehicle_Count"],
    bins=30,
    kde=True
)

plt.title("Vehicle Count Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Traffic_Speed_kmph"],
    bins=30,
    kde=True
)

plt.title("Traffic Speed Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["Traffic_Density_vpkm"],
    bins=30,
    kde=True
)

plt.title("Traffic Density Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Congestion_Level",
    y="Vehicle_Count"
)

plt.title("Vehicle Count vs Congestion")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Congestion_Level",
    y="Traffic_Speed_kmph"
)

plt.title("Traffic Speed vs Congestion")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Congestion_Level",
    y="Traffic_Density_vpkm"
)

plt.title("Traffic Density vs Congestion")

plt.show()
```
